# 3. NumPy Demonstration

```
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

## 3.1 Working with Arrays

```
values = np.array([4, 7, 9, 12, 15])


statistics = {
    "Statistic": ["Mean", "Minimum", "Maximum"],
    "Value": [
        np.mean(values),
        np.min(values),
        np.max(values)
    ]
}

import pandas as pd
pd.DataFrame(statistics)
```
