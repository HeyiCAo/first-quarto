# NumPy Demonstration

```
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

As shown in @fig-numpy-wave, NumPy can generate and transform numerical data that can then be visualized clearly.
